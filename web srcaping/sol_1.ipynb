```python
from bs4 import BeautifulSoup

import requests
import pandas as pd
```

```python
import os 
os.makedirs("web_s" , exist_ok = True)
print("create:-" , os.path.abspath("web_s"))
```

```python
page_count = 1
while True:
    url = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(url)
    if res.status_code == 404:
        break
    soup = BeautifulSoup( res.text , "html.parser" )
    quotes = soup.select("div.quote")
    if not quotes:
        break
    with open (f"web_s/quotes_pages{page_count}.html" , "w" , encoding = "utf-8" ) as f:
        f.write(res.text)
    print(f"pages number {page_count} is sucessfully fetch " )
    page_count += 1
```

```python
import shutil
if os.path.exists("web_s"):
    shutil.rmtree("web_s")
```

```python
with open("web_s/quotes_pages1.html" , "r" , encoding = "utf-8") as f:
    html_content = f.read()
soup = BeautifulSoup( html_content , "html.parser" )
soup 
```

```python
header = soup.select("div.quote")
life_quotes = []
for i in header:
    all_tags = []
    for j in i.select(".tags .tag"):
        all_tags.append(j.get_text())
    if "life" in all_tags:
        text = i.select_one("span.text").get_text()
        author = i.select_one("small.author").get_text()
        life_quotes.append([text, author])
print(life_quotes)
```

```python
df_life = pd.DataFrame(life_quotes, columns=["Quote", "Author"])
```
