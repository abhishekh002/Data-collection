```python
from bs4 import BeautifulSoup

import requests
import pandas as pd
```

```python
page_count = 1
while True:
    res = requests.get(f"https://quotes.toscrape.com/page/{page_count}/")
    if res.status_code == 404:
        break
    soup = BeautifulSoup(res.text, "html.parser")
    quotes = soup.select("div.quote")
    if not quotes:
        break
    print(f"Page {page_count}: {len(quotes)} quotes")

    with open(f"web_scraping/quotes{page_count}.html" , "w" , encoding="utf-8") as f:
              f.write(res.text)
    
              
    page_count += 1
```
