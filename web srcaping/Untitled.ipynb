```python
import numpy as np
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

```python
url = "https://www.scrapethissite.com/pages/forms/"
res = requests.get(url)
```

```python
soup = BeautifulSoup(res.text, "html.parser")
```

```python
soup
```

```python
for row in soup.find_all('tr'):
    print(row)
```

```python
for name in soup.find_all('td', class_='name'):
    print(name.text.strip())
```

```python
for row in soup.find_all('tr', class_='team'):
    name = row.find('td', class_='name').text.strip()
    year = row.find('td', class_='year').text.strip()
    wins = row.find('td', class_='wins').text.strip()
    losses = row.find('td', class_='losses').text.strip()
    gf = row.find('td', class_='gf').text.strip()
    ga = row.find('td', class_='ga').text.strip()
    
    print(f"Team: {name} | Year: {year} | Wins: {wins} | Losses: {losses} | GF: {gf} | GA: {ga}")
```

```python
import pandas as pd

data = []
for row in soup.find_all('tr', class_='team'):
    name = row.find('td', class_='name').text.strip()
    year = row.find('td', class_='year').text.strip()
    wins = row.find('td', class_='wins').text.strip()
    losses = row.find('td', class_='losses').text.strip()
    gf = row.find('td', class_='gf').text.strip()
    ga = row.find('td', class_='ga').text.strip()
    
    data.append({
        'Team': name,
        'Year': year,
        'Wins': wins,
        'Losses': losses,
        'GF': gf,
        'GA': ga
    })

# Convert to DataFrame
df = pd.DataFrame(data)

# Save to CSV
df.to_csv('hockey_teams.csv', index=False)
print("Saved successfully to hockey_teams.csv!")
```
